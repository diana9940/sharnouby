## 4) Forecast
1. **Moving Average (3 شهور)** — متوسط آخر 3 شهور (الأبسط).
2. **Seasonal Naive × Growth** — نفس الشهر السنة اللي فاتت × معدل النمو.
3. **Seasonal Index + Trend** — نشيل الموسمية، نرسم خط اتجاه، ونرجّع الموسمية.

**التقييم (Backtest):** نخبّي آخر 6 شهور (أبريل–سبتمبر 2026)، نتنبأ بيها، ونقارن بالحقيقي بـ MAPE.

```python
seg = sales.groupby([sales.Order_Date.dt.to_period('M'), 'Segment']).Net_Sales_EGP.sum().unstack()
H = 6
def ma3(y, h):  return np.repeat(y[-3:].mean(), h)
def snaive_growth(y, h):
    g = np.clip(y[-12:].sum() / y[-24:-12].sum(), .7, 1.5); return np.array([y[-12 + k] * g for k in range(h)])
def seasonal_trend(y, h):
    n = len(y); mon = np.arange(n) % 12; idx = np.array([y[mon == m].mean() for m in range(12)]); idx /= idx.mean()
    des = y / idx[mon]; b, a = np.polyfit(np.arange(n)[-24:], des[-24:], 1)
    return np.array([(a + b * (n + k)) * idx[(n + k) % 12] for k in range(h)])
METHODS = {'Moving Avg (3m)': ma3, 'Seasonal Naive x Growth': snaive_growth, 'Seasonal Index + Trend': seasonal_trend}
mape = lambda a, f: np.mean(np.abs(a - f) / a) * 100
bt = {}
for s_ in seg.columns:
    y = seg[s_].values; tr, te = y[:-6], y[-6:]
    bt[s_] = {m: mape(te, f(tr, 6)) for m, f in METHODS.items()}
bt = pd.DataFrame(bt).round(1); display(bt.style.highlight_min(color='#bfe8d6').format('{:.1f}%'))
best = bt.idxmin().to_dict(); print('Best method per segment:', best)
```

```python
last = seg.index[-1]; fidx = pd.period_range(last + 1, periods=H, freq='M'); rows = []
fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
for ax, s_ in zip(axes, seg.columns):
    y = seg[s_].values; m = best[s_]; fc = METHODS[m](y, H); err = bt.loc[m, s_] / 100
    test_fc = METHODS[m](y[:-6], 6)
    ax.plot(seg.index.to_timestamp(), y, color='#1d5fe0', lw=2, label='Actual')
    ax.plot(seg.index[-6:].to_timestamp(), test_fc, color='#16a37a', ls=':', lw=2, label='Backtest (hidden 6m)')
    ax.plot([seg.index[-1].to_timestamp()] + list(fidx.to_timestamp()), [y[-1]] + list(fc), color='#e0912a', ls='--', marker='o', lw=2.5, label=f'Forecast ({m})')
    ax.fill_between(fidx.to_timestamp(), fc * (1 - err), fc * (1 + err), color='#e0912a', alpha=.2, label=f'± backtest error ({err:.0%})')
    ax.yaxis.set_major_formatter(M1); ax.set_title(f'{s_} — net sales (EGP)'); ax.legend(fontsize=8)
    for p, v in zip(fidx, fc): rows.append(dict(Month=p.to_timestamp(), Segment=s_, Method=m, Forecast=v, Lower=v * (1 - err), Upper=v * (1 + err)))
save('14_forecast')
fc_tbl = pd.DataFrame(rows); fc_tbl['Month'] = fc_tbl.Month.dt.strftime('%Y-%m')
display(fc_tbl.pivot(index='Month', columns='Segment', values='Forecast').assign(Total=lambda d: d.sum(axis=1)).div(1e6).round(1).rename(columns=lambda c: c + ' (EGP M)'))
```

## 5) What-if — لو قلّلنا خصومات الجملة؟


```python
w = sales[(sales.Channel == 'Wholesale') & (sales.Discount_Pct > .30)]
gp_now = (w.Net_Sales_EGP - w.Estimated_COGS_EGP).sum(); gp_cap = (w.Gross_Sales_EGP * .70 - w.Estimated_COGS_EGP).sum()
print(f'Wholesale lines with discount > 30%: {len(w):,} ({len(w) / (sales.Channel == "Wholesale").sum():.0%} of wholesale lines)')
print(f'Their gross profit now: EGP {gp_now/1e6:.1f}M | with 30% cap (same volume): EGP {gp_cap/1e6:.1f}M')
res = pd.DataFrame({'Volume lost on these lines': [0, .05, .10, .20, .30]})
res['Gross profit (EGP M)'] = [(gp_cap * (1 - l)) / 1e6 for l in res.iloc[:, 0]]; res['Change vs now (EGP M)'] = res['Gross profit (EGP M)'] - gp_now / 1e6
display(res.round(1)); print(f'Break-even: profit stays higher unless more than {1 - gp_now / gp_cap:.0%} of that volume is lost.')
```
